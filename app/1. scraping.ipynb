```python
from fed_extract_text import extract_pdf_links, extract_text_from_pdf

links_text_path = os.path.join(path_data, "links_text.txt")

try:
    with open(links_text_path, "r", encoding="utf-8") as f:
        links_text = json.load(f)
except FileNotFoundError:
    links_text = []

processed_urls = {item["url"] for item in links_text}


def is_press_release(url: str) -> bool:
    url_lower = url.lower()
    return "fomcpre" in url_lower or "pressreleases" in url_lower


def get_relevant_pdf_links(url: str) -> list[str]:
    pdf_links = extract_pdf_links(url, root="https://www.federalreserve.gov")
    return [
        link
        for link in pdf_links
        if "presc" in link.lower() or "longer" in link.lower()
    ]


def extract_result(url: str) -> dict:
    if not is_press_release(url):
        return extract_text_from_link(url)

    print(f"URL {url} é uma página de comunicado de imprensa, extraindo links de PDF...")
    pdf_links = get_relevant_pdf_links(url)
    print(f"Links de PDF encontrados: {pdf_links}")

    if not pdf_links:
        return extract_text_from_link(url)

    if len(pdf_links) > 1:
        # Seleciona somente o primeiro link de PDF encontrado
        print(f"Encontrados múltiplos links de PDF para a URL: {url}. Selecionando o primeiro link: {pdf_links[0]}")

        if "LongerRunGoals" in pdf_links[0]:
            pdf_links = [pdf_links[0]]
        else:
            raise ValueError(f"Encontrados múltiplos links de PDF para a URL: {url}")

    pdf_link = pdf_links[0]
    print(f"Extraindo texto do PDF {pdf_link}...")

    metadata = extract_text_from_link(url, title_pubdate=True)

    return {
        "text": extract_text_from_pdf(pdf_link),
        "title": metadata["title"],
        "publication_date": metadata["publication_date"],
    }


max_new_items = 99999
processed_now = 0

for i, url in enumerate(links, start=1):
    print(f"Processando URL {i}/{len(links)}: {url}")

    if url in processed_urls:
        print(f"URL {url} já foi processada, pulando...")
        continue

    result = extract_result(url)

    if result.get("text") is None:
        continue

    title = result.get("title") or ""
    if "implementation note" in title.lower():
        result["text"] = "NOT EXTRACTED - IMPLEMENTATION NOTE"

    links_text.append(
        {
            "url": url,
            "title": result.get("title"),
            "text": result.get("text"),
            "publication_date": result.get("publication_date"),
        }
    )
    processed_urls.add(url)

    with open(links_text_path, "w", encoding="utf-8") as f:
        json.dump(links_text, f, indent=2, ensure_ascii=False)

    processed_now += 1
    if processed_now >= max_new_items:
        break

```
